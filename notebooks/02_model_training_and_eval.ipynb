# HealthGuard AI: 4-Algorithm Model Benchmarking & Evaluation

**Objective**: Benchmark the 4-algorithm ML pipeline (K-Means Patient Segmentation, Recursive Feature Elimination, Logistic Regression Baseline, and XGBoost Ensemble Engine) on 10,000 patient records.

---

## 1. Environment Setup & Pipeline Ingestion

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sys.path.append(os.path.abspath('..'))
from src.model import FourAlgorithmPipeline

dataset_path = os.path.abspath('../data/readmission_dataset.csv')
df = pd.read_csv(dataset_path)
print(f"Loaded dataset: {df.shape[0]} records.")
```

## 2. Execute 4-Algorithm Pipeline Training

```python
pipeline = FourAlgorithmPipeline(dataset_path=dataset_path, n_clusters=4, n_rfe_features=15)
metrics = pipeline.train(df)

print("--- 4-ALGORITHM EVALUATION RESULTS ---")
print(f"Logistic Regression AUC-ROC : {metrics['logistic_regression']['auc_roc']:.4f} | Accuracy: {metrics['logistic_regression']['accuracy']*100:.2f}%")
print(f"XGBoost Predictive Engine AUC: {metrics['xgboost']['auc_roc']:.4f} | Accuracy: {metrics['xgboost']['accuracy']*100:.2f}%")
```

## 3. RFE Feature Ranking & XGBoost Feature Weight Distribution

```python
df_imp = pd.DataFrame({
    'Feature': list(metrics['xgboost']['feature_importances'].keys()),
    'Importance': list(metrics['xgboost']['feature_importances'].values())
}).sort_values('Importance', ascending=False).reset_index(drop=True)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=df_imp,
    x='Importance',
    y='Feature',
    hue='Feature',
    palette='crest',
    legend=False,
    dodge=False,
 )
plt.title('Feature Importance Comparison')
plt.xlabel('Importance Score')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```
